```python
import pandas as pd
```

```python
df_data=pd.read_csv("/content/healthcare_dataset.csv")
```

```python
df_data
```

Validation Data

```python
df_data.head()
```

```python
df_data.tail()
```

pre processing

```python
df_columns=list(df_data.select_dtypes('object').columns)
```

```python
df_columns
```

```python
from sklearn.preprocessing import LabelEncoder

Encode=LabelEncoder()
```

```python
for column_name in df_columns:
  df_data[column_name]=Encode.fit_transform(df_data[column_name])
```

```python
df_data
```

Null value Handling

```python
df_data.isnull().sum()
```

Split Input and Output

```python
x=df_data.drop('Medication',axis=1)
y=df_data['Medication']
```

Machine Learning flow

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
y_test
```

Model creation

```python
from sklearn.linear_model import LinearRegression

LR=LinearRegression()

```

```python
LR.fit(x_train,y_train)
```

```python
LR.predict(x_test)
```

```python
LR.score(x_train,y_train)
```

```python
LR.score(x_test,y_test)
```

LogisticRegression

```python
from sklearn.linear_model import LogisticRegression

LR=LogisticRegression()
```

```python
LR.fit(x_train,y_train)
```

```python
LR.predict(x_test)
```

```python
LR.score(x_train,y_train)*100
```

```python
LR.score(x_test,y_test)*100
```

Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

DT=DecisionTreeClassifier()
```

```python
DT.fit(x_train,y_train)
```

```python
DT.predict(x_test)
```

```python
DT.score(x_train,y_train)*100
```

```python
DT.score(x_test,y_test)*100
```

RandomForest

```python
from sklearn.ensemble import RandomForestClassifier

RF=RandomForestClassifier()
```

```python
RF.fit(x_train,y_train)
```

```python
RF.predict(x_test)
```

```python
RF.score(x_train,y_train)*100
```

```python
RF.score(x_test,y_test)*100
```
